```python
def foo1(x=7):
    return x ** 0.5

def foo2(x=3, y=5):
    if x > y:
        return x
    else:
        return y

def foo3(x=1729):
    d = 2
    myfactors = []
    while x > 1:
        if x % d == 0:
            myfactors.append(d)
            x = x / d
        else:
            d = d + 1
    return myfactors

print("foo1:", foo1())
print("foo2:", foo2())
print("foo3:", foo3())
```

## Debugging the Function

I used the debugger to check the position and codon as the function ran.
The first pair was (0, 'AUG'). The next pair was (4, 'AAU').

I found that the function moved forward four bases instead of three.
This skipped a base and produced the wrong amino acids.

I changed `i = i + 4` to `i = i + 3`.
I also changed `<` to `<=` so the function includes a final complete codon.

```python
def get_amino_acids(mRNA):
    i = 0
    aa_sequence = []

    while (i + 3) <= len(mRNA):
        codon = mRNA[i:i + 3]
        aa = genetic_code[codon]

        if aa == "Stop":
            break

        aa_sequence.append(aa)
        i = i + 3

    return "".join(aa_sequence)

print(get_amino_acids(test_mRNA))
```

```python
from pathlib import Path

repo = Path("/Users/sahilghani/IntroBiolComp-2026")
input_file = repo / "Python/DataFiles/Turkey_transcripts_15_coding.fasta"
output_folder = repo / "Practicals/W6"
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "Turkey_proteins_15.fasta"

# Read the DNA sequences
records = []
header = None
parts = []

with open(input_file) as file:
    for line in file:
        line = line.strip()
        if not line:
            continue

        if line.startswith(">"):
            if header is not None:
                records.append((header, "".join(parts)))
            header = line
            parts = []
        else:
            parts.append(line)

if header is not None:
    records.append((header, "".join(parts)))

assert len(records) == 15, "Expected 15 sequences"

# Translate each sequence and update its header
proteins = []

for header, dna in records:
    assert header.endswith("gbkey=CDS"), "Unexpected header ending"
    new_header = header[:-len("gbkey=CDS")] + "gbskey=AA"
    mRNA = dna.upper().replace("T", "U")
    protein = get_amino_acids(mRNA)
    proteins.append((new_header, protein))

# Save the protein sequences as FASTA
with open(output_file, "w") as file:
    for header, protein in proteins:
        file.write(header + "\n")
        for start in range(0, len(protein), 60):
            file.write(protein[start:start + 60] + "\n")

assert all(header.endswith("gbskey=AA") for header, _ in proteins)
assert all(protein for _, protein in proteins)

print("Sequences read:", len(records))
print("Proteins saved:", len(proteins))
print("All protein headers end in gbskey=AA.")
print("Saved:", output_file)
```

```python
from pathlib import Path
import subprocess
import sys

folder = Path("/Users/sahilghani/IntroBiolComp-2026/Practicals/W6")

script = '''import pickle
from pathlib import Path

repo = Path(__file__).resolve().parents[2]

with open(repo / "Python/genetic_code.pickle", "rb") as file:
    genetic_code = pickle.load(file)


def get_amino_acids(mRNA):
    amino_acids = []

    for i in range(0, len(mRNA) - 2, 3):
        codon = mRNA[i:i + 3]
        aa = genetic_code[codon]
        if aa == "Stop":
            break
        amino_acids.append(aa)

    return "".join(amino_acids)


def read_fasta(path):
    records = []
    header = None
    parts = []

    with open(path) as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if header is not None:
                    records.append((header, "".join(parts)))
                header = line
                parts = []
            else:
                parts.append(line)

    if header is not None:
        records.append((header, "".join(parts)))

    return records


if __name__ == "__main__":
    assert get_amino_acids("AUGGAAUUCUCGCUCUGAAGGUAA") == "MEFSL"
    assert get_amino_acids("AUG") == "M"

    input_file = repo / "Python/DataFiles/Turkey_transcripts_15_coding.fasta"
    output_file = Path(__file__).resolve().parent / "Turkey_proteins_15.fasta"
    records = read_fasta(input_file)
    assert len(records) == 15

    translated = 0
    with open(output_file, "w") as file:
        for header, dna in records:
            assert header.endswith("gbkey=CDS")
            new_header = header[:-len("gbkey=CDS")] + "gbskey=AA"
            protein = get_amino_acids(dna.upper().replace("T", "U"))

            file.write(new_header + "\\n")
            if not dna:
                print("Missing DNA:", header.split()[0])
                file.write("\\n")
                continue

            assert protein, "Nonempty DNA produced an empty protein"
            translated += 1
            for start in range(0, len(protein), 60):
                file.write(protein[start:start + 60] + "\\n")

    print("Records saved:", len(records))
    print("Proteins translated:", translated)
    print("Saved:", output_file)
'''

script_file = folder / "translate_turkey.py"
script_file.write_text(script, encoding="utf-8")

result = subprocess.run(
    [sys.executable, str(script_file)],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
else:
    print("Script ran successfully.")
```
